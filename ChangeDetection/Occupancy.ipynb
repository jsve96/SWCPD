### RULSIF

```python
import numpy as np
import pandas as pd
from densratio import densratio
#from sklearn.preprocessing import StandardScaler
from matplotlib import pyplot as plt
from tqdm import tqdm

# --- 1. Define the Sliding Window Function ---

def calculate_rulsif_scores(
    time_series_data, # (N_timesteps, N_features) numpy array
    window_size,
    step_size,
    alpha=0.1
):
    """
    Applies RuLSIF using a two-window sliding approach for change detection.

    Args:
        time_series_data (np.ndarray): The multivariate time series data.
        window_size (int): The size of the reference (x1) and test (x2) windows.
        step_size (int): The number of steps to slide the windows forward.
        alpha (float): The relative density parameter for RuLSIF (0.1 is standard).

    Returns:
        tuple: (list of window center indices, list of RPD scores)
    """
    N, D = time_series_data.shape
    rpd_scores = []
    window_indices = []
    
    # The comparison starts after the initial reference window is established
    start_index = window_size
    
    # Loop to slide the two windows (Reference and Test)
    # The total window span is 2 * window_size. The first score is registered 
    # at the end of the Test Window (which is the potential change-point)
    for i in tqdm(range(start_index, N - window_size, step_size)):
        
        # 1. Define the Reference Window (x1) - Past data
        # From i - window_size to i
        x1 = time_series_data[i - window_size : i, :]
        
        # 2. Define the Test Window (x2) - Current data
        # From i to i + window_size
        x2 = time_series_data[i : i + window_size, :]
        
        # Check if windows are of the correct size (safety check)
        if x1.shape[0] != window_size or x2.shape[0] != window_size:
            continue
            
        try:
            # 3. Apply RuLSIF
            # x1 is the numerator (p(x)), x2 is the denominator (q(x))
            densratio_obj = densratio(
                x1, x2, 
                alpha=alpha,
                sigma_range="auto", 
                lambda_range="auto", verbose=False
            )
            
            # The alpha-relative PE divergence (RPD) is a measure of dissimilarity
            # between the distributions p(x) and q(x).
            score = densratio_obj.alpha_PE
            rpd_scores.append(score)
            
            # The score is logged at the end of the Test Window
            window_indices.append(i + window_size) 
            
        except Exception as e:
            # Handle cases where optimization might fail (rare with good data)
            print(f"RuLSIF error at index {i}: {e}")
            rpd_scores.append(np.nan)
            window_indices.append(i + window_size)
        #print(densratio_obj.alpha_PE)  
    return np.array(window_indices), np.array(rpd_scores)

signal = (Occupancy-Occupancy.mean())/Occupancy.std()
signal = signal.values.astype(np.float32)

detection_RULIF = calculate_rulsif_scores(signal[:,0][:,np.newaxis],20,step_size=2)
```
